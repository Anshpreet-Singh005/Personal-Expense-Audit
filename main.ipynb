# Personal Expense Audit

This notebook uses Python and Pandas to review a personal expense dataset.

## Objectives
- Inspect the records, columns, and data types
- Check for missing and duplicate records
- Review transaction amounts and identify the largest expenses
- Standardize category names
- Compare category totals, transaction counts, and average amounts
- Examine spending by day and month
- Summarize total expenditure and average transaction amount

```python
import pandas as pd
```

## 1. Load the Dataset

The expense data is stored in a CSV file. We will load it into a Pandas DataFrame for analysis.

```python
df = pd.read_csv("personal_expense_audit.csv")
```

### Initial Data Inspection

First, we display the complete dataset to get an initial understanding of the available records.

```python
df
```

### Dataset Structure

We use `info()` to examine the number of entries, column names, data types, and non-null values.

```python
df.info()
```

### Statistical Summary

The `describe()` method provides summary statistics for the numerical columns in the dataset.

```python
df.describe()
```

### Missing Value Check

We check each column for missing values to determine whether the dataset requires any missing-data treatment.

```python
df.isna().sum()
```

### Duplicate Record Check

Count exact duplicate rows for review. Identical transactions can be legitimate, so check them before deciding whether to remove any.

```python
df.duplicated().sum()
```

### Viewing the Dataset

We inspect the first and last few records to verify the structure and contents of the dataset.

```python
df.head()
```

```python
df.tail()
```

## 2. Expense Analysis

Review transaction amounts and identify the largest expense.

### Identifying the Largest Transaction

We identify the transaction with the highest expense amount and display its complete details.

```python
df[df['Amount'] == df['Amount'].max()][['Date', 'Category', 'Description','Amount']]
```

## 3. Category Cleaning

Before analyzing spending by category, we first examine the existing category values for inconsistencies such as different capitalization or unnecessary spaces.

### Inspecting Category Values

Check the unique category values for differences in capitalization and leading or trailing spaces before standardizing them.

```python
df['Category'].unique()
```

### Standardizing Category Names

The category values contain inconsistent formatting. We standardize them by removing unnecessary spaces and using consistent capitalization.

```python
df['Category'] = df['Category'].str.capitalize().str.strip()
```

### Verify the Cleaning

We check the unique category values again to confirm that the categories have been standardized successfully.

```python
df['Category'].unique()
```

## 4. Category Analysis

We first count the number of transactions belonging to each expense category.

```python
df['Category'].value_counts()
```

### Total Spending by Category

We calculate the total amount spent in each category using `groupby()` and `sum()`.

```python
category_totals = df.groupby("Category")["Amount"].sum()
category_totals.sort_values(ascending= False)
```

### Highest-Spending Category

The totals above compare expenditure across categories. This cell returns the category with the greatest total spending.

```python
category_totals.idxmax()
```

### Average Transaction Amount by Category

Compare the average transaction amount within each category. A category’s average can differ from its total because categories contain different numbers of transactions.

```python
df.groupby("Category")["Amount"].mean()
```

## 5. Spending Over Time

Convert transaction dates to datetime values so spending can be grouped by day and month.

```python
df['Date'] = pd.to_datetime(df['Date'])
```

### Daily Spending

Group transactions by date to calculate the total amount spent on each day.

```python
daily_spending = df.groupby('Date')['Amount'].sum()
```

```python
daily_spending
```

### Highest-Spending Day

Identify the date with the largest daily total and the amount spent that day.

```python
daily_spending.idxmax()
```

```python
daily_spending.max()
```

### Monthly Spending

Sum transaction amounts by calendar month. The current dataset covers 2026, so grouping by month number is sufficient here.

```python
df.groupby(df['Date'].dt.month)['Amount'].sum()
```

### Top Three Categories by Total Spending

Rank categories by total expenditure and display the three highest.

```python
df.groupby("Category")["Amount"].sum().sort_values(ascending=False).head(3)
```

## 6. Overall Summary

Summarize total expenditure and the average transaction amount for the full dataset.

```python
print("Total Expenditure:", df["Amount"].sum())
print("Average Transaction:", df["Amount"].mean())
```
